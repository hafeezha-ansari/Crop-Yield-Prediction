```python
# Data handling
import pandas as pd
import numpy as np

# Data visualization
import matplotlib.pyplot as plt
import seaborn as sns

# Machine Learning - splitting data
from sklearn.model_selection import train_test_split

# Machine Learning - regression model
from sklearn.ensemble import RandomForestRegressor

# Machine Learning - evaluation metrics
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
```

```python
import pandas as pd

df = pd.read_csv('crop_yield_dataset.csv')

df.head()
```

```python
# Check the number of rows and columns
print("Number of rows and columns:", df.shape)
```

```python
# Display column names
print("\nColumn names:")
print(df.columns)
```

```python
# Check data types
print("\nData types:")
print(df.dtypes)
```

```python
# Check for missing values
print("\nMissing values:")
print(df.isnull().sum())
```

```python
# Check the different crop types in the dataset
print("Number of different crops:", df['Crop'].nunique())
```

```python
print("\nCrop types:")
print(df['Crop'].unique())
```

```python
print("\nNumber of records for each crop:")
print(df['Crop'].value_counts())
```

```python
# Display basic statistical information
df.describe()
```

```python
# Select input features
X = df[
    [
        'Temperature_C',
        'Rainfall_mm',
        'Soil_Moisture_pct',
        'Soil_pH',
        'Nitrogen_kg_ha',
        'Phosphorus_kg_ha',
        'Potassium_kg_ha'
    ]
]

# Select target variable
y = df['Yield_ton_ha']

# Display the first 5 rows of X
print("Input Features (X):")
display(X.head())

# Display the first 5 target values
print("\nTarget Variable (y):")
display(y.head())
```

```python
# Split the dataset into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

# Check the size of each dataset
print("Training data:", X_train.shape)
print("Testing data:", X_test.shape)
print("Training target:", y_train.shape)
print("Testing target:", y_test.shape)
```

```python
# Create the Random Forest Regression model
model = RandomForestRegressor(
    n_estimators=100,
    random_state=42
)

# Train the model using the training data
model.fit(X_train, y_train)

print("Model training completed successfully!")
```

```python
# Make predictions on the testing data
y_pred = model.predict(X_test)

# Display the first 10 predictions
print("First 10 predicted crop yields:")
print(y_pred[:10])
```

```python
# Calculate evaluation metrics

mae = mean_absolute_error(y_test, y_pred)
mse = mean_squared_error(y_test, y_pred)
rmse = np.sqrt(mse)
r2 = r2_score(y_test, y_pred)

# Display the results
print("Model Evaluation Results")
print("------------------------")
print("MAE  :", round(mae, 4))
print("MSE  :", round(mse, 4))
print("RMSE :", round(rmse, 4))
print("R²   :", round(r2, 4))
```

```python
# Plot Actual vs Predicted Crop Yield

plt.figure(figsize=(8, 6))

plt.scatter(y_test, y_pred)

# Perfect prediction line
plt.plot(
    [y_test.min(), y_test.max()],
    [y_test.min(), y_test.max()],
    linestyle='--'
)

plt.xlabel("Actual Crop Yield (ton/ha)")
plt.ylabel("Predicted Crop Yield (ton/ha)")
plt.title("Actual vs Predicted Crop Yield")

plt.show()
```

```python
# Get feature importance from the Random Forest model
importance = model.feature_importances_

# Create a DataFrame for better visualization
feature_importance = pd.DataFrame({
    'Feature': X.columns,
    'Importance': importance
})

# Sort from highest to lowest importance
feature_importance = feature_importance.sort_values(
    by='Importance',
    ascending=False
)

# Display the results
print("Feature Importance:")
display(feature_importance)
```

```python
# Plot feature importance

plt.figure(figsize=(9, 6))

plt.barh(
    feature_importance['Feature'],
    feature_importance['Importance']
)

plt.xlabel("Importance")
plt.ylabel("Feature")
plt.title("Feature Importance in Crop Yield Prediction")

plt.gca().invert_yaxis()

plt.show()
```

Key Insight:

Soil moisture was the most important feature in the Random Forest model, contributing approximately 51.2% of the model's feature importance. Nitrogen and rainfall were the next most important factors.
